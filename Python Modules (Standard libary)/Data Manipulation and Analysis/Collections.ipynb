#### **improving code readability**

**namedtuple()**

To acces elements in a tuple we might need to ise the indices but in some cases to count or get the indicies of certain element in a tuple might not be easy or a little bit of a work. `namedtuple` allows us to access these values uding the **dot notation**.

In short, the `namedtuple()` is a factory function that allows you yo create tuple subclasses with named fields. Thet are hashable (can work as dictionary keys and can be stored in sets), support slicing and indexing, backward compatible with regular tuples and have similar memory usage to regular tuple.


```python
from collections import namedtuple
```

```python
Point = namedtuple("point", "x y z")

point = Point(1, 2, 3)
point
```

```python
print(point.x)
print(point.y)
```

```python
# they are not mutable
point.x = 9 # will throw an attribute error
```

```python
print(issubclass(Point, tuple))
```

Though just like regular tuples, namedtuples are immutable the ob ject they store or hold don't neccesarily have to be immutable.

```python
Person = namedtuple("Person", "name children")
john = Person("John Doe", ["sanjay", "craige"])

id(john.children) 
```

```python
type(john.children)
```

we see that the element of the namedtuple in our example is a list and this is mutable. 


```python
john.children.append("nina")
```

```python
john
```

```python
hash(john)
```

we observe from above that namedtuples or tuples with mutable values aren't hashable

They are several ways to instantiate a namedtuple some of which include:

```python
# using a list of string for the field names
Point = namedtuple("Point", ["x", "y"])
print(Point)
print(Point(2, 4))
```

```python
# using strings with comma-seperated field names
Point = namedtuple("Point", "x, y")
print(Point)
print(Point(2, 4))
```

```python
# using generator expression for the filed names
Point = namedtuple("Point", (field for field in "xy"))
print(Point)
print(Point(2, 4))
```

we could even use a set, namedtuples has some cool optional arguements:

* `rename`
* `defaults`
* `module`


if we set `rename = True` it set's all invalid fiels and duplicate fields automatically to positional names such as `_3`, `_2` etc

```python
WeirdPoint = namedtuple("WeirdPoint", ["x", "y", "x", "class"], rename=True)
print(WeirdPoint._fields) # class is invalid here because it is a reserved python keyword
```

```python
# custom invalids
invalids = {"foo", "bar", "baz"}

def safe_namedtuple(name, fields):
    # Replace any "invalid" with None so rename=True will rename them
    fixed = [f if f not in invalids else None for f in fields]
    return namedtuple(name, fixed, rename=True)

MyTuple = safe_namedtuple("MyTuple", ["x", "foo", "y", "bar"])
print(MyTuple._fields)
```

the next arguement defaukt is used to set defaukt field values incase user of the named tuple don't enter these fields.

```python
Dev = namedtuple(
    "Developer",
    "name level language",
    defaults=["junior", "Python"]
)

Dev(john)
```

### **deque: building Queues and stacks**

Thre word deque is pronounced "deck" and it stands for double-ended queue.

In python the append and pop operation on the beginning or left-side of the list is O(n) time complexity in the worst case (not frequent).

which can be expensive for very large lists and we ould want something that is of constant time (O(n) time complexity).

On the other, the  append and pop operation on the right side of a list are normally efficient (O(1)) except for those cases where python needs to reallocate memory to grow the underlying list for accepting new items which happens when working with a lot of data.

The python deque are implemented as doubly linked list. Making it possible to append and pop elements from both sides of the deque in a stable and efficient manner (also O(1)). This is why we use deques for creating stacks and queues

The Queue follows tge FIFO fashion (First in - First out). Adding an item to any end of a queue is called **enqueue** and removing an item from any end of the queue is called **dequeue**.

```python
from collections import deque

new_queue = deque()
new_queue
```

```python
# assume people join the queue (from behind of course)
new_queue.append("John")
new_queue.append("Jane")
new_queue.append("Sanjay")

new_queue
```

```python
# say people already got there food and are leaving the queue

new_queue.popleft()
```

```python
new_queue.popleft()
```

```python
new_queue.popleft()
```

```python
new_queue.popleft()
```

The deque takes two optional arguements which are the `initializer` and `maxlen`

```python
food_queue = deque(["John", "Jane", "Sanjay"], maxlen=3)
food_queue
```

```python
food_queue.appendleft("Craig")
food_queue
```

```python
food_queue.appendleft("Smith")
```

```python
food_queue
```

#### **Extend on existing queue**

```python
lucky_numbers = deque(range(5))
lucky_numbers
```

```python
lucky_numbers.extend([5, 6, 7, 8, 9])
lucky_numbers
```

we will observe that wend we extendleft a queue, the first elemnt in our list (we choose to estend into the queue), the first element goes in first then we continue in that other till the end.

```python
lucky_numbers.extendleft([-1, -2, -3, -4, -5])
lucky_numbers
```

**Note:** The `,pop()` method in deque does'nt support an arbitrary indicies arguemnet and the `.insert()` method is supporteed by the deque in the fashion `.insert(index, item)`

```python
lucky_numbers.insert(2, -10)
lucky_numbers
```

```python
# other methods ion deque

# create a copy of a deque
lucky_numbers_copy = lucky_numbers.copy()
print(f"copy of lucky_numbers {lucky_numbers_copy}")

# count number of apppearance of item x in deque
print(f"10 appears {lucky_numbers.count(-10)} in lucky numbers")

# remove the first occurence of an element in a deque
lucky_numbers.remove(-10)
print("removing -10 from lucky numbers")
print(lucky_numbers)

# clear the deque
print("clearing lucky numbers")
lucky_numbers.clear()
print(lucky_numbers)
```
