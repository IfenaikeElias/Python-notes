**namedtuple()**

To acces elements in a tuple we might need to ise the indices but in some cases to count or get the indicies of certain element in a tuple might not be easy or a little bit of a work. `namedtuple` allows us to access these values uding the **dot notation**.

In short, the `namedtuple()` is a factory function that allows you yo create tuple subclasses with named fields. Thet are hashable (can work as dictionary keys and can be stored in sets), support slicing and indexing, backward compatible with regular tuples and have similar memory usage to regular tuple.


```python
from collections import namedtuple
```

```python
Point = namedtuple("point", "x y z")

point = Point(1, 2, 3)
point
```

```python
print(point.x)
print(point.y)
```

```python
# they are not mutable
point.x = 9 # will throw an attribute error
```

```python
print(issubclass(Point, tuple))
```

They are several ways to instantiate a namedtuple some of which include:

```python
# using a list of string for the field names
Point = namedtuple("Point", ["x", "y"])
print(Point)
print(Point(2, 4))
```

```python
# using strings with comma-seperated field names
Point = namedtuple("Point", "x, y")
print(Point)
print(Point(2, 4))
```

```python
# using generator expression for the filed names
Point = namedtuple("Point", (field for field in "xy"))
print(Point)
print(Point(2, 4))
```

we could even use a set, namedtuples has some cool optional arguements:

* `rename`
* `defaults`
* `module`


if we set `rename = True` it set's all invalid fiels and duplicate fields automatically to positional names such as `_3`, `_2` etc

```python
WeirdPoint = namedtuple("WeirdPoint", ["x", "y", "x", "class"], rename=True)
print(WeirdPoint._fields) # class is invalid here because it is a reserved python keyword
```

```python
# custom invalids
invalids = {"foo", "bar", "baz"}

def safe_namedtuple(name, fields):
    # Replace any "invalid" with None so rename=True will rename them
    fixed = [f if f not in invalids else None for f in fields]
    return namedtuple(name, fixed, rename=True)

MyTuple = safe_namedtuple("MyTuple", ["x", "foo", "y", "bar"])
print(MyTuple._fields)
```

the next arguement defaukt is used to set defaukt field values incase user of the named tuple don't enter these fields.

```python
Dev = namedtuple(
    "Developer",
    "name level language",
    defaults=["junior", "Python"]
)

Dev(john)
```

#### **Exploring some features of named tuples**

**Converting Named Tuples into Dictionaries with `._asdict()`**

```python
Person = namedtuple("Person", "name age height")
nick = Person("jane", 25, 1.75)
nick._asdict()
```

**replacing the values of a namedtuple using `._replace`**

```python
Person = namedtuple("Person", "name age height")
jane = Person("Jane", 25, 1.75)

# After Jane's birthday
jane = jane._replace(age=26)
jane
```
